## Теория

Рассмотрим задачу бинарной классификации, причем $\mathscr{Y} = \{+1, -1\}$. Пусть так же $\widehat{y}$ &mdash; некоторый классификатор, предсказывающий <u><i>степень принадлежности</i></u> классу. При этом решающее правило имеет вид
 $f(x) = \text{sign}\left(\widehat{y}(x)\right)$.
В задачах 1-4 рассматриваем логистическую функцию потерь:
$$\mathcal{L}(y, z) = \log \big(1 + \exp(-yz) \big).$$

### <span style="display:inline-block;padding:6px 13px;border-radius:999px;background:linear-gradient(135deg,#D4470E 0%,#C7324E 55%,#B316B8 100%);color:#FFFFFF;font-size:16px;font-weight:700;line-height:1.2;">Задача 1</span>

Рассмотрим градиентный бустинг с логистической функцией потерь. Выпишите для градиентного спуска формулу для вектора сдвигов предсказаний и задачу поиска новой базовой модели по их приближению.



1. Вычисляем градиенты по обучающей выборке:

$\mathcal{L}(y, z) = \log(1 + \exp(-yz)$ - дана

тогда посчитаем градиент по предсказаниям $z$:

$\frac{\partial \mathcal{L}(y, z)}{\partial z} = \frac{-y \cdot exp(-yz)}{1 + exp(-yz)}$

Тогда $\tilde{g}_i^t = \frac{\partial \mathcal{L}(Y_i, \hat{y}_{t - 1}(X_i))}{\partial z} = \frac{-Y_i \cdot exp(-Y_i \cdot \hat{y}_{t - 1}(X_i))}{1 + exp(-Y_i \cdot \hat{y}_{t - 1}(X_i))}$

А вектор сдвигов предсказаний (то есть куда мы еще хотим сдвинуть наше суммарное предсказание чтобы итог стал похожим на $Y_i$) это есть $-\tilde{g}_i^t$.

2. Обучить новую модел по НАШЕЙ ФУНКЦИИ ПОТЕРЬ по выборке $(X_i, -\tilde{g}_i^t)$
$b_t$ модель на $t$-ом последовательном шаге обучается на выборке $(X_i, -\tilde{g}_i^t)$, предсказывать направление антиградиента, чтобы допредсказывать всё то, что не допредсказывали прошлые модели вместе взятые

как обучать ее будем:

Пусть $\mathcal{F}$ - это множество базовых моделей

$b_t(x) \in \mathcal{F}$, оно должно предсказывать $-\tilde{g}_i^t$, поскольку в нашем случае таргеты это числа, то и градиент то же число, а ни что лучше не помогает нам понять, насколько точно мы предсказали число (задача регрессии), как MSE.

$b_t(x) = \text{arg} \text{min}_{b \in \mathcal{F}} \sum_{i = 1}^n \text{MSE}(b(X_i), -\tilde{g}_i^t)$

$b_t(x) = \text{arg} \text{min}_{b \in \mathcal{F}} \sum_{i = 1}^n (b(X_i) -(-\tilde{g}_i^t))^2$

### <span style="display:inline-block;padding:6px 13px;border-radius:999px;background:linear-gradient(135deg,#D4470E 0%,#C7324E 55%,#B316B8 100%);color:#FFFFFF;font-size:16px;font-weight:700;line-height:1.2;">Задача 2</span>

Предположим, модель градиентного бустинга $\widehat{y}_{t - 1}$ уже построена.

**1.** Выпишите вид функционала ошибки $Q(\widehat{y}_t) = \sum\limits_{i = 1}^n \mathcal{L}\left(Y_i, \widehat{y}_t(x_i)\right)$ для логистической функции потерь. Одинаковый ли вклад вносят разные объекты в ошибку при построении новой базовой модели $b_t$?



**Ответ:**

вид функционала ошибки $Q(\widehat{y}_t) = \sum\limits_{i = 1}^n \mathcal{L}\left(Y_i, \widehat{y}_t(x_i)\right)$ для логистической функции потерь такой:

$Q(\widehat{y}_t) = \sum\limits_{i = 1}^n \log(1 + \exp(-Y_i \cdot \widehat{y}_t(x_i)))$



То есть у нас сейчас используется LogitBoost, логарифмическая гладкая функция для аппроксимации пороговой функции ошибки



Рассмотрим поподробнее $Q(\widehat{y}_t)$, чтобы понять какой **вклад** каждый объект вносит в функцию ошибки $Q$.

Начнём с того, что сделаем тот же трюк что и для AdaBoost (для экспоненциальной функции потерь) на лекции - воспользоваться свойством экспоненты:

$Q(\widehat{y}_t) = \sum\limits_{i = 1}^n \log(1 + \exp(-Y_i \cdot (\widehat{y}_{t - 1}(x_i) + \eta_t \cdot b_t(X_i))))$

Далее пользуемся свойствами экспоненты и получаем:

$Q(\widehat{y}_t) = \sum\limits_{i = 1}^n \log(1 + \exp(-Y_i \cdot \widehat{y}_{t - 1}(x_i)) \cdot \exp(-Y_i \cdot \eta_t \cdot b_t(X_i)))$

Вот эту вот вещь мы знаем: $\exp(-Y_i \cdot \widehat{y}_{t - 1}(x_i))$ - это константа, поскольку все предыдущие $t - 1$ модель нашего прекрасного ансамбля уже построены. И больше того: $M_i = Y_i \cdot \widehat{y}_{t - 1}(x_i)$, отступ внутрь класса (то есть если предсказание и реальная метка таргета одного класса +1 оба или -1 оба, то отступ > 0, причем можно сказать что чем больше отступ тем дальше внутри класса мы находимся, это можно в терминах проекций из алгема объяснить)

Поскольку логарифм это возрастающая функция, то конечно, как видно из полученной формулы разные объекты вносят разный вклад.

Теперь чуть подробнее:

Чтобы явно понять, какие веса дают объекты, я предлагаю разложить логарифм по Формуле Тэйлора

Перепишем функцию ошибки, подставив $M_i$

$Q(\widehat{y}_t) = \sum_{i = 1}^n \log(1 + \exp(-M_i) \cdot \exp(-Y_i \cdot \eta_t \cdot b_t(X_i)))$

И обозначим $\Delta x_i = -Y_i \cdot \eta_t \cdot b_t(X_i)$

Тогда разложение в Тэйлора выглядит так, где $\mathcal{L}_i$ - лосс на $i$-ом объекте:

$\mathcal{L}_i(\Delta x_i) = \mathcal{L}_i(0) + \frac{\partial \mathcal{L}_i}{\partial \Delta x_i} |_{\Delta x_i = 0} \cdot \Delta x_i + o(\Delta x_i)$

В силу того, что $\mathcal{L}_i(0) = \log(1 + \exp(-M_i))$:

Мы уже считали в первой задаче частную производную лосса по $z$, воспользуемся: $\frac{\partial}{\partial \Delta x_i} \log(1 + \exp(-M_i) \cdot \exp(\Delta x_i)) = \frac{\exp(-M_i) \cdot \exp(\Delta x_i)}{1 + \exp(-M_i) \cdot \exp(\Delta x_i)}$

Тогда в подстановке получим:

$\frac{\partial \mathcal{L}_i}{\partial \Delta x_i} |_{\Delta x_i = 0} = \frac{\exp(-M_i)}{1 + \exp(-M_i)} = \frac{1}{\exp(M_i) + 1}$

Тогда ошибка перепишется так:

$Q(\widehat{y}_t) = \sum_{i = 1}^n \log(1 + \exp(-M_i)) + \sum_{i = 1}^n \frac{1}{1 + \exp(M_i)} \cdot (-Y_i \cdot \eta_t \cdot b_t(X_i)) + \sum_{i = 1}^n o(\eta_t \cdot b_t(X_i))$

То есть $Q(\widehat{y}_t) = \text{const} - \eta_t \cdot \sum_{i = 1}^n w_i \cdot Y_i \cdot b_t(X_i) + \sum_{i = 1}^n o(\eta_t \cdot b_t(X_i))$

$w_i = \frac{1}{1 + \exp(M_i)} = \sigma(-M_i)$ -- вот такие веса объектов получаются

**Анализируем**

1. $M_i >> 0$

то есть модель очень уверена в своем правильном предсказаниии

Тогда $w_i$ маленькое. То есть объект, в случае если он правильно классифицирован, он учитывается с маленьким весом в функции ошибки.

2. $M_i << 0$

то есть модель неверно предсказала

Тогда $w_i$ около 1 значение, то есть объект учитывается со своим "настоящим весом" единичкой в функции ошибки.

**То есть $w_i$ - мера важности объекта $X_i$ на $t$-ой итерации.** Что я имею под этим ввиду:

При выборе оптимального $b_t$, $b_t(X_i)$ очень похож на $-\tilde{g}_i^t$, а значит минимизируется и ошибка, в которой стоит сумма таких градиентов приравненная к нулю (для того, чтобы минимизировать ошибку). То есть минимизация ошибки $Q$ и подбор оптимальной новой модели неразрывно связаны. А значит $w_i$ можно интепретировать так: насколько большое внимание $b_t(X_i)$ обращать внимание на объект $X_i$, то есть на сколько сильно для него надо исправлять значение $\hat{y}_{t - 1}(X_i)$ - то что ранее понапредсказывали в купе базовые модели вместе.

То есть вклад разные объекты вносят конечно разный, и мы его явно посчитали и выразили через отступ $M_i$.


**2.** Посмотрите на формулу для вектора сдвигов, полученную в предыдущей задаче. Как она выражается через отклики на объектах обучающей выборки? Одинаковый ли вклад вносят разные объекты в формирование вектора сдвигов при построении новой базовой модели $b_t$?


**Ответ:**

Формула вектора сдвигов полученная в предыдущей задаче:

$\tilde{g}_i^t = \frac{\partial \mathcal{L}(Y_i, \hat{y}_{t - 1}(X_i))}{\partial z} = \frac{-Y_i \cdot exp(-Y_i \cdot \hat{y}_{t - 1}(X_i))}{1 + exp(-Y_i \cdot \hat{y}_{t - 1}(X_i))} = \frac{-Y_i}{1 + exp(Y_i \cdot \hat{y}_{t - 1}(X_i))} = -Y_i \cdot \sigma(-M_i)$

где $M_i$ - все так же margin (отступ) объекта, $M_i = Y_i \cdot \hat{y}_{t - 1}(X_i))$, зависит от откликов объектов на уже обученных $t - 1$ модели.

А вектор сдвигов предсказаний (то есть куда мы еще хотим сдвинуть наше суммарное предсказание чтобы итог стал похожим на $Y_i$) это есть $-\tilde{g}_i^t = Y_i \cdot \sigma(-M_i)$.

Мы выше показали, что разные объекты выборки учитываются в функционале ошибки с разным весом $w_i = |-\tilde{g}_i^t| = \sigma(-M_i)$, так как $|Y_i| = 1$ (так как таргеты это +1, -1). А минимизация функционала ошибки и подбор оптимальной базовой модели на шаге $t$ неразрывно связано. $b_t$ учится предсказывать антиградиенты $-\tilde{g}_i^t$, а на них зачение ошибки как раз минимизируется по определению градиента.

Так что вклад объектов в построение модели $b_t$ не одинаков, потому что значения компонент векторов сдвигов отличаются:

1. $M_i >> 0$

то есть модель очень уверена в своем правильном предсказаниии

Тогда $-\tilde{g}_i^t$ маленькое. То есть объект, в случае если он правильно классифицирован, он мало учитывается новой моделью $b_t$, в том плане что $b_t$ должна исправить совсем чуть-чуть ошибку на этом объекте. А значит вклад этого объекта в построение $b_t$ и формирование вектора сдвигов невелик.

2. $M_i << 0$

то есть модель неверно предсказала

Тогда $-\tilde{g}_i^t$ около 1 значение по модулю, то есть объект учитывается новой моделью $b_t$ со своим "настоящим весом" единичкой, то есть $b_t$ должна неплохо так исправить ошибку которую допустил уже построенный ансамбль на этом объекте. Поэтому влияние этого $X_i$ объекта на $b_t$ и формирование вектора сдвигов ощутимо.

**3.**  На лекции было показано, что для экспоненциальной функции есть проблема: базовый классификатор может настраиваться только на шумовые объекты. Наблюдается ли такая проблема у логистической функции потерь? Как она ведет себя на объектах с сильно отрицательным, околонулевым, сильно положительным отсупами?

**Ответ:**



Действительно, у AdaBoost есть такая проблема, за счет того, что вес с которым объект $X_i$ учитывается в ошибке (суммарный лосс по объектам), экспоненциально зависит от отступа $M_i$ с минусом, то есть $-M_i$. Если в данных есть шум, или конкретно для этого одногт объекта $X_i$ была случайная ошибка в данных, которая привела к неверной классификации объекта, то его $M_i < 0$ причем скорее всего сильно меньше нуля. Это значит что такой "шумовой" объект будет давать вклад в ошибку $exp(-M_i)$ - огромный (так как $-M_i > 0$). За счет этого, чтобы минимизировать эту ошибку базовые модели (особенно деревья) могут сильно переобучиться.

У нас вес каждого объекта в фцнкцию ошибки (суммарный лосс по объектам) с логистической функцией потерь такой: $w_i = \sigma(-M_i)$.

Во-первых, сразу оговоримся, что $\sigma$ : $\mathbb{R} \to [0, 1]$, то есть вес объекта ограничен единичкой и экспоненциально возрасти он так не может, то есть у логситической функции потерь такой проблемы нет.

Во-вторых посмотрим на поведение веса $w_i = \sigma(-M_i)$ при различных значения отступа:

1. $M_i << 0$ - сильноотрицательный

то $-M_i >> 0$ - сильноположительный, а значит $w_i$ примерно около единички (вспоминаем график сигмоиды). То есть вес в этом случае максимальный. Почему это логично? Если отступ сильно отрицательный, это значит, что объект был неправильно классифицирован, на нем ансамбль ошибся. Надо эту ошибку исправить. Поэтому дадим объекту максимально возможный (в рамках данной логистической функции потерь вес), то есть единичку. Это круче чем у AdBoost. Модель $b_t$ будет при обучении обращать пристальное внимание на этот объект $X_i$, но при этом не будет такого жесткого переобучения как у AdaBoost.

2. $M_i$ околонулевой

то $-M_i$ -  тоже околонулевой. Вспоминая график сигмроиды понимаем, что $w_i = \sigma(-M_i)$ примерно 0.5. То есть когда $M_i$ околонуля, то ансамбль наш не очень то уверен в своем предсказании для этого объекта, а значит дадим ему вес $0.5$, чтобы при обучении новая модель $b_t$ на него внимание, но не так сильно как на объекты с весом один. Обратила внимание в том смысле, что $b_t$ учится на $-\tilde{g}_i^t$, а веса объектов и обучение $b_t$ неразрывно связаны: $w_i = |-\tilde{g}_i^t|$ (это мы поняли выше)

3. $M_i >> 0$ - сильноположительный

то $-M_i << 0$ - сильноотрицательный, вспоминая график сигмоиды понимаем, что в таком случае $w_i = \sigma(-M_i)$ около нуля, то есть вес такому объекту присвоится очень маленький. Сильноположительный отступ означает что модель довольно уверена в своем предсказании, и новой модели $b_t$ для этого объекта $X_i$ уже можно "почти" ничего не исправлять (ошибку предыдущего ансамбля на шаге $t - 1$).

Рассмотрим градиентный бустинг над решающими деревьями с логистической функции потерь. После построения дерева будем делать перенастройку в листьях дерева.

**1.** Выпишите оптимизационную задачу для коэффициентов $\gamma_{tj}$ &mdash; новых ответов в листьях.




**Ответ:**







Запишем по аналогии с лекцией метод Ньютона (решение задачи подбора оптимального предсказания в каждом листе одного решающего дерева)

В $j$-ом листе дерева $b_t$ делается такая перенастройка предсказания:

$F_j^t(\gamma) = \sum_{X_i \in R_j} \log(1 + \exp(-Y_i \cdot [\hat{y}_{t-1} + \gamma])) \to \text{min}_{\gamma}$

**2.** Решите полученную задачу сделав один шаг метода Ньютона из начального приближения $\gamma_{tj} = 0$, что соответствует отсутствию базовой модели $b_t$.

**Ответ:**

Задачу мы поставили, так что тогда формула пересчета по методу Ньютона такова:

$\gamma_{tj} = \gamma_{tj} - \frac{\frac{\partial F_j^t(0)}{\partial \gamma}}{\frac{\partial^2 F_j^t(0)}{\partial \gamma^2}}$

(откуда взялась такая формула пересчета - смотри чат ИИ)

Осталось посчитать:

1. $\frac{\partial F_j^t(0)}{\partial \gamma}$ - эту штуковину мы в предыдущих задачах уже посчитали:

как обычно $M_i = Y_i \cdot \hat{y}_{t-1}$.

$\frac{\partial F_j^t(0)}{\partial \gamma} = \sum_{X_i \in R_j} \frac{-Y_i}{1 + \exp(Y_i \cdot \hat{y}_{t-1})}$

2. $\frac{\partial^2 F_j^t(0)}{\partial \gamma^2}$

$\frac{\partial^2 F_j^t(0)}{\partial \gamma^2} = \sum_{X_i \in R_j} \frac{Y_i^2 \cdot \exp(Y_i \cdot \hat{y}_{t-1})}{(1 + \exp(Y_i \cdot \hat{y}_{t-1}))^2}$

Таргеты $Y_i$ принимают значения $+1$ или $-1$, то $Y_i^2 = 1$, и формулу можно немного упростить, убрав $Y_i^2 = 1$

То есть $\frac{\partial^2 F_j^t(0)}{\partial \gamma^2} = \sum_{X_i \in R_j} \frac{\exp(Y_i \cdot \hat{y}_{t-1})}{(1 + \exp(Y_i \cdot \hat{y}_{t-1}))^2}$

3. Итоговый шаг метода Ньютона

Изначально (как на лекции): $\gamma_{tj} = 0$

$\gamma_{tj} = 0 - \frac{\frac{\partial F_j^t(0)}{\partial \gamma}}{\frac{\partial^2 F_j^t(0)}{\partial \gamma^2}} = -\frac{\sum_{X_i \in R_j} \frac{-Y_i}{1 + \exp(Y_i \cdot \hat{y}_{t-1})}}{\sum_{X_i \in R_j} \frac{\exp(Y_i \cdot \hat{y}_{t-1})}{(1 + \exp(Y_i \cdot \hat{y}_{t-1}))^2}}$

1. Посмотрите, как меняется качество модели в зависимости от значений гиперпараметров и сделайте выводы по результатам.

Исследуем сначала зависимость качества моделей двух (Random Forest Regressor, GradientBoostingRegressor) от значения `n_estimators`:

Сперва обучим лес на `n_estimators` деревьях

```python
from sklearn.ensemble import RandomForestRegressor, RandomForestClassifier, GradientBoostingRegressor
from sklearn.metrics import mean_squared_error, r2_score
from typing import Dict, Optional, Callable
import random

random_forest = RandomForestRegressor(n_estimators=400, random_state=RANDOM_STATE)
random_forest.fit(X_train, y_train)

def plot_some_lines(
    lines: Dict[str, np.ndarray],
    xlabel: str,
    estrimator_range: Optional[np.ndarray] = None,
    title: Optional[str] = None,
    metric_name: str = 'MSE',
):
    '''
    Отрисовывает линии из словаря lines

    Параметры:
    lines (Dict[str, np.ndarray]): словарь линий, ключ - label, значение - линия
    xlabel (str): название оси X
    estrimator_range (np.ndarray): значения на оси X
    metric_name (Optional[str]): имя метрики
    title (str): заголовок графика
    '''

    with sns.axes_style("darkgrid"):
        plt.figure(figsize=(12, 6))
        plt.title(title, fontsize=20)

        for name, line in lines.items():
            if estrimator_range is not None:
                plt.plot(estrimator_range, line, label=name, lw=3)
            else:
                plt.plot(line, label=name, lw=3)
        plt.grid(ls=':')
        plt.xlabel(xlabel)
        plt.ylabel(f'Метрика {metric_name}')
        plt.legend(fontsize=14)
        plt.show()


def cum_metric(
    model: RandomForestRegressor,
    metric: Callable[[np.ndarray, np.ndarray], float],
    x_test: np.ndarray,
    y_test: np.ndarray,
    shuffle: bool = False,
    binary_ths: float = 0.5,
):
    '''
    Считает значения метрики metric в зависимости от количества деревьев
    в обученной модели model метом композиции деревьев.

    Параметры:
    model (RandomForestRegressor): модель RandomForest
    metric (Callable[[np.ndarray, np.ndarray], float]): метрика
    x_test (np.ndarray): тестовая выборка
    y_test (np.ndarray): ответы на тестовой выборке
    shuffle (bool): надо ли перемешивать деревья
    binary_ths (float): порог вероятности для истинного класса
                       в случае задачи классификации

    Возвращает: кумулятивные значения метрики
    '''

    # Поведение для классификатора и регрессии будет немного отличаться
    is_classifier = isinstance(model, RandomForestClassifier)

    if isinstance(model, RandomForestRegressor):
        # Считаем предсказания по каждому дереву отдельно
        predictions_by_estimators = [
            est.predict_proba(x_test)[:, 1] if is_classifier else est.predict(x_test)
            for est in model.estimators_
        ]

        if shuffle:
            random.shuffle(predictions_by_estimators)

        numerator = np.cumsum(predictions_by_estimators, axis=0)
        n = len(predictions_by_estimators)
        denominator = (np.arange(n) + 1)[:, np.newaxis]
        cumpred = numerator / denominator

        if is_classifier:
            cumpred = cumpred > binary_ths

        cumacc = [metric(y_test, pred) for pred in cumpred]

    else: # Gradient Boosting
        predictions_history = list(model.staged_predict(x_test))
        cumacc = [metric(y_test, pred) for pred in predictions_history]
    return np.array(cumacc)

#MSE SCORE
mse_score = lambda y_true, y_pred: mean_squared_error(y_true, y_pred)

# СЧИТАЕМ МЕТРИКИ ДЛЯ СЛУЧАЙНОГО ЛЕСА
cum_mse_train_forest = cum_metric(random_forest, mse_score, X_train, y_train)
cum_mse_test_forest = cum_metric(random_forest, mse_score, X_test, y_test)
estrimator_range_forest = np.arange(random_forest.n_estimators) + 1

# ОБУЧАЕМ БУСТИНГ
grad_boost_model = GradientBoostingRegressor(n_estimators=400, random_state=RANDOM_STATE)
grad_boost_model.fit(X_train, y_train)

# СЧИТАЕМ МЕТРИКИ ДЛЯ БУСТИНГА
cum_mse_train_boost = cum_metric(grad_boost_model, mse_score, X_train, y_train)
cum_mse_test_boost = cum_metric(grad_boost_model, mse_score, X_test, y_test)
estrimator_range_boost = np.arange(grad_boost_model.n_estimators) + 1

plot_some_lines(
    lines={
        'Тренировочная выборка RandomForest': cum_mse_train_forest,
        'Тестовая выборка RandomForest': cum_mse_test_forest,
        'Тренировочная выборка GradBoost': cum_mse_train_boost,
        'Тестовая выборка GradBoost': cum_mse_test_boost,
    },
    xlabel='n_estimators',
    estrimator_range=estrimator_range_forest, # оно одинаковое
    title=f'Зависимость MSE от n_estimators',
    metric_name=f'MSE',
)
```

Видно, что с ростом количества деревьев в ансамбле как в случае RandomForest, так и в случае градиентного бустинга, метрика MSE падает и с какого-то момента выходит на плато.

Особенности:

видно что на трэйн значение MSE в случае RandomForest упало ниже, чем для GradBoost. Потому что, по умолчанию у градиентного бустинга max_depth = 3, то есть обучаются короткие пни, а значит они не могут сильно переобучиться, в то время как у RandomForest по умолчаниб стоит значение max_depth = None, то есть без ограничения на глубину, а значит каждое дерево в случайном лесу может переобучиться, и на трэйн выдавать очень маленький (относительно задачи) MSE.

При этом видно, что на старте при `n_estimators` до 50, RandomForest резко выходит на плато при `n_estimators` = 15 примерно, в то время как градиентный бустинг плавно выходит на плато (я предполагаю, что это связано, что в градиентном бустинге предсказания всех моделей ансамбля берутся с весами， и из-за этого мы не все предсказания делим на кол-во деревьев `n_estimators`, как в Random Forest просто усредняется, а всех берем с умным весом, вторая причина - в градиентном бустинге обучаются пни max_depth=3, а в Random Forest глубина деревьев по умолчанию не ограничена).

Поэтому градиентный бустинг выходит на плато поздно, при `n_estimators` = 200

Теперь зафиксируем n_estimators равным 400 и построим график зависимости качества двух типов моделей от гиперпараметра - максимальная глубина дерева (`max_depth`) [от 1 до 15];

```python
from typing import Sequence, Any
import copy
import tqdm

def get_train_and_test_mse(
    param_name: str,
    grid: Sequence[Any],
    other_params_dict: Dict[str, Any] = {},
):
    '''
    Функция для оценки точности классификации MSE
    для заданных значений параметра param_name

    Параметры:
    param_name (str): название параметра, который собираемся варьировать
    grid (Sequence[Any]): сетка значений параметра
    other_params_dict (Dict[str, Any]): словарь со значениями остальных параметров
    '''

    train_mse_forest, test_mse_forest, train_mse_boosting, test_mse_boosting = [], [], [], []
    params_dict = copy.copy(other_params_dict)

    for param_value in tqdm.tqdm(grid):
        params_dict.update({param_name: param_value})
        estimator_forest = RandomForestRegressor(**params_dict, random_state=RANDOM_STATE, n_jobs=-1)
        estimator_forest.fit(X_train, y_train)
        train_mse_forest.append(mean_squared_error(y_train, estimator_forest.predict(X_train)))
        test_mse_forest.append(mean_squared_error(y_test, estimator_forest.predict(X_test)))

        estimator_boosting = GradientBoostingRegressor(**params_dict, random_state=RANDOM_STATE)
        # ВАЖНО: так как бустинг работает последовательно а не паралельно как RandomForest обучает свои деревья, то n_job=-1 такого параметра нет у бустинга, потому что ничего распаралеливать не надо
        estimator_boosting.fit(X_train, y_train)
        train_mse_boosting.append(mean_squared_error(y_train, estimator_boosting.predict(X_train)))
        test_mse_boosting.append(mean_squared_error(y_test, estimator_boosting.predict(X_test)))

    return train_mse_forest, test_mse_forest, train_mse_boosting, test_mse_boosting


def plot_dependence(
    param_name: str,
    grid: Sequence[Any] = range(2, 20),
    other_params_dict: Dict[str, Any] = {}
):
    '''
    Функция для отображения графика зависимости MSE
    от значения параметра c названием param_name

    Параметры:
    param_name (str): название параметра, который собираемся варьировать
    grid (Sequence[Any]): сетка значений параметра
    other_params_dict (Dict[str, Any]): словарь со значениями остальных параметров
    '''

    train_mse_forest, test_mse_forest, train_mse_boosting, test_mse_boosting =\
        get_train_and_test_mse(
            param_name, grid, other_params_dict
        )

    plot_some_lines(
        lines={
            'Тренировочная выборка RandomForest': train_mse_forest,
            'Тестовая выборка RandomForest': test_mse_forest,
            'Тренировочная выборка GradBoost': train_mse_boosting,
            'Тестовая выборка GradBoost': test_mse_boosting,
        },
        xlabel=param_name,
        estrimator_range=grid,
        title=f'Зависимость MSE от {param_name}',
        metric_name=f'MSE',
    )
```

Почему так по-разному?



То есть:

* Random Forest (Bagging) борется со смещением модели, то есть он разрешает деревьям быть глубокими и переобучаться. А потом за счет стохастики (max_features), то есть попарной нескоррелированности деревьев, и за счет усреднения пропадает дисперсия итогового моделя (а одно дерево конечно неустойчиво, то есть дисперсия предсказаний большая, поскольку одно дерево просто переобучилось под выборку и обобщающая способность у него никакая)
* Gradient Boosting заходит с другой стороны, он борется с дисперсией модели, то есть он берет много маленьких деревьев (не переобученных, а значит с маленькой дисперсией предсказаний, в силу простоты модели у нее весьма устойчивые предсказания), а смещение итогового ансамбля (да одиночные такие деревья смещены, поскольку их предсказания далеки от реальности, они слишком "слепы" и просты) нивелируется за счет определения градиентного бустинга - каждое следующее дерево в ансамбле исправляет ошибки предыдущего, то есть смещение к концу построения всего ансамбля последовательно уходит.

Поэтому из-за того, что они работают по разному им нужны разные гиперпараметры. GradientBoosting хочет пни, а RandomForest хочет деревья.

Теперь рассмотрим зависимость метрики MSE для GradientBoostingRegressor от learning_rate (это из градиентного спуска к нам пришло, когда мы следующее дерево предсказываем так чтобы оно исправляло ошибки предыдущих, то есть шло по напраывлению антиградиента лосса по предсказаниям)

```python
def get_train_and_test_mse(
    param_name: str,
    grid: Sequence[Any]
):
    '''
    Функция для оценки точности классификации MSE
    для заданных значений параметра param_name

    Параметры:
    param_name (str): название параметра, который собираемся варьировать
    grid (Sequence[Any]): сетка значений параметра
    '''

    train_mse, test_mse = [], []
    params_dict = {}

    for param_value in tqdm.tqdm(grid):
        params_dict.update({param_name: param_value})
        estimator_boosting = GradientBoostingRegressor(**params_dict, random_state=RANDOM_STATE)
        estimator_boosting.fit(X_train, y_train)
        train_mse.append(mean_squared_error(y_train, estimator_boosting.predict(X_train)))
        test_mse.append(mean_squared_error(y_test, estimator_boosting.predict(X_test)))
    return train_mse, test_mse


def plot_dependence(
    param_name: str,
    grid: Sequence[Any] = range(2, 20)
):
    '''
    Функция для отображения графика зависимости MSE
    от значения параметра c названием param_name

    Параметры:
    param_name (str): название параметра, который собираемся варьировать
    grid (Sequence[Any]): сетка значений параметра
    '''

    train_mse, test_mse = get_train_and_test_mse(
        param_name, grid
    )

    plot_some_lines(
        lines={
            'Тренировочная выборка': train_mse,
            'Тестовая выборка': test_mse,
        },
        xlabel=param_name,
        estrimator_range=grid,
        title=f'Зависимость MSE для GradBoost от {param_name}',
        metric_name=f'MSE',
    )


plot_dependence("learning_rate", [0.0, 0.001, 0.003, 0.005, 0.01, 0.02, 0.03, 0.05, 0.08, 0.1, 0.15, 0.2, 0.3, 0.5, 1.0, 1.25, 1.5, 2.0])
```

```python
import time
def get_time_statistics(model, model_name, X_train, y_train, X_test, y_test):
    start_fit_time = time.time()
    model.fit(X_train, y_train)
    end_fit_time = time.time()

    fitting_time = end_fit_time - start_fit_time

    times_per_object = []

    for i in range(len(X_test)):
        object_i = np.array(X_test[i : i + 1, :])

        start_pred_time = time.perf_counter()
        model.predict(object_i)
        end_pred_time = time.perf_counter()

        times_per_object.append(end_pred_time - start_pred_time)

    times_per_object = np.array(times_per_object)
    mean_time_per_obj = np.mean(times_per_object)
    lower_quantile = np.quantile(times_per_object, 0.025)
    upper_quantile = np.quantile(times_per_object, 0.975)

    plt.figure(figsize=(10, 6))
    plt.hist(times_per_object, bins=50, label='Распределение времени предсказания (сек)')
    plt.axvline(mean_time_per_obj, color='purple', linestyle='--', label=f'Среднее: {mean_time_per_obj} сек')

    plt.axvline(lower_quantile, color='red', linestyle='dotted', label=f'2.5% квантиль: {lower_quantile} сек')
    plt.axvline(upper_quantile, color='red', linestyle='dotted', label=f'97.5% квантиль: {upper_quantile} сек')

    plt.title(f'Распределение времени предсказания (сек) на один объект для {model_name}')
    plt.xlabel('Время выполнения (сек)')
    plt.ylabel('Количество объектов')
    plt.legend()
    plt.show()

    print(f"Время обучения модели {model_name} = {fitting_time}")


# Проводим сравнение
forest_optimal = RandomForestRegressor(**random_search_forest.best_params_, random_state=RANDOM_STATE)
boosting_optimal = GradientBoostingRegressor(**random_search_boost.best_params_, random_state=RANDOM_STATE)

get_time_statistics(forest_optimal, "RandomForest", X_train, y_train, X_test, y_test)
get_time_statistics(boosting_optimal, "GradBoost", X_train, y_train, X_test, y_test)
```

У бустинга есть метод staged_predict. Он выдает предсказания композиции первых $t$ моделей бустинга. Используя этот метод,нарисуем для каждого $t \in \{1, ..., 10\}$ два графика:

* График предсказаний первых 10 моделей и истинных ответов;
* График ошибки на шаге $t \in \{1, ..., 10\}$ и предсказаний $t$-й компоненты бустинга.

```python
# сортируем наши данные по возрастанию иначе график получится некорректным
x_raw = X_train[:, most_imp_feature_ind]
sort_idx = np.argsort(x_raw)
X_train_single = x_raw[sort_idx].reshape(-1, 1)
y_train_sorted = y_train[sort_idx]

boosting_optimal.fit(X_train_single, y_train_sorted) # заново обучили

# Копируем код с семинара
grid = np.linspace(X_train_single.min(),X_train_single.max(), 500).reshape(-1, 1)

staged = list(boosting_optimal.staged_predict(X_train_single))
staged_grid = list(boosting_optimal.staged_predict(grid))

# Начальная модель в sklearn — константа (среднее таргета), а не ноль
init_train = boosting_optimal.init_.predict(X_train_single)
init_grid = boosting_optimal.init_.predict(grid)

for i in range(10):
    previous_predict = staged[i-1] if i > 0 else init_train
    previous_grid = staged_grid[i-1] if i > 0 else init_grid

    error = y_train_sorted - previous_predict

    fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(13, 5))

    # Первый график
    scatter1 = ax1.scatter(X_train_single, y_train_sorted, label='Обучающие данные', s=15, alpha=0.7)
    line1, = ax1.plot(grid, staged_grid[i], color='red', linewidth=2, label='Предсказание ансамбля')
    ax1.set_xlabel('Признак')
    ax1.set_ylabel('Таргет')
    ax1.set_title(f"Итерация {i+1}")
    ax1.legend(fontsize=11, loc='best')

    # Второй график
    scatter2 = ax2.scatter(X_train_single, error, label='Ошибка предыдущего шага', s=15, alpha=0.7)
    line2, = ax2.plot(grid, staged_grid[i] - previous_grid,
                     color='red', linewidth=2, label=f'Коррекция {i+1}')
    ax2.set_xlabel('Признак')
    ax2.set_ylabel('Ошибка')
    ax2.set_title(f"Итерация {i+1}")
    ax2.legend(fontsize=11, loc='best')

    plt.tight_layout()
    plt.show()
```

**4.** **Функция лосса**

Для задачи регрессии разработано достаточно много функций потерь, в том числе с дополнительными свойствами робастности. Один такой пример — функция потерь Хьюбера, она же Huber loss. Суть функции в том, что на небольших отклонениях она работает как $\large L_2$, а с заранее заданного порога, начинает работать как $\large L_1$. Это позволяет уменьшить вклад выбросов и следующих за ними квадратично-больших ошибок на общий вид функции, при этом не акцентируя внимание на мелких неточностях и отклонениях.


$R(x)$ &mdash; функция потерь Хьюбера, определяемая как
$$R(x) = \frac{x^2}{2} I\left\{|x| < c\right\} + c \left(|x| - \frac{c}{2}\right)I\left\{|x| > c\right\}.$$

Обучите модель градиентного бустинга используя данный лосс сравните результат с MSE.

```python
from sklearn.metrics import mean_absolute_error

boosting_mse = GradientBoostingRegressor(**random_search_boost.best_params_, loss="squared_error")
boosting_mse.fit(X_train, y_train)
y_pred_mse = boosting_mse.predict(X_test)
print(f"MSE для boosting_mse = {mean_squared_error(y_pred_mse, y_test)}")
print(f"MAE для boosting_mse = {mean_absolute_error(y_pred_mse, y_test)}")

boosting_huber = GradientBoostingRegressor(**random_search_boost.best_params_, loss="huber")
boosting_huber.fit(X_train, y_train)
y_pred_h = boosting_huber.predict(X_test)
print(f"MSE для boosting_huber = {mean_squared_error(y_pred_h, y_test)}")
print(f"MAE для boosting_huber = {mean_absolute_error(y_pred_h, y_test)}")
```

так как Хьюбер это комбинация параболы, и модуля $|x|$ начиная с некоторого момента, то весьма понятно желание сравнить MSE, MAE как критерий качества для обоих подходов

Видно, что MSE для бустинга который обучался с функцией потерь MSE, у него MSE меньше итоговое, а у Huber'а больше.

Так же у бустинга который обучался на Huber, MAE немного повыше, чем у бустинга который обучился на MSE.

Проблема лосса Хбюбера в том, что появляется еще один гиперпараметр $c$, который надо подобрать так, чтобы модель улучшила свое качество. Если бы я не использовала дефолтный $c$, а попробовала разный, возможно получилось бы добиться тако же как для бустинг , который обучался на MSE, какчества а может даже и лучшего качества
