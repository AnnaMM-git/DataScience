### <span style="display:inline-block;padding:6px 13px;border-radius:999px;background:linear-gradient(135deg,#D4470E 0%,#C7324E 55%,#B316B8 100%);color:#FFFFFF;font-size:16px;font-weight:700;line-height:1.2;">Задача 1</span>

Пусть обучающая выборка $(X_1, Y_1), ..., (X_n, Y_n)$ такова, что
* объекты $X_1, ..., X_n$ одномерны и имеют распределение $\mathcal{N}(0, \sigma^2)$;
* отклик получается по правилу $Y_i = X_i^2 + \varepsilon_i$, где $\varepsilon_i$ независимы, имеют нулевое среднее и не зависят от $X_i$.

Пусть также для объекта $X$ отклик $Y$ получен по аналогичному правилу, причем $X$ и $Y$ не зависят от обучающей выборки.

Для МНК-модели $\widehat{y}(x) = \widehat{\theta} x$ выпишите подробное bias-variance разложение со всеми тремя компонентами: смещение, разброс, шум. Компоненту, отвечающую за разброс, разрешается не доводить до конца. Если в выражении присутствует дисперсия $\varepsilon_i$, то так и пишите: $\mathsf{D}\varepsilon_i$.

По теореме с лекции мы знаем, что если:
* $\hat{y}$ - это предсказание нашей модели,
* $y = f(X, \varepsilon)$ - истинная зависимость данных
то тогда среднеквадратичная ошибка модели считается так:

$\text{MSE}(\hat{y}) = \mathbb{E}(\hat{y}(x) - \mathbb{E} f(x, \varepsilon))^2 + \mathbb{D}(\hat{y}(x)) + \mathbb{D} f(x, \varepsilon)$

Так же из этой теоремы были два следствия: объединю их результаты в одно:

1) Если истинная зависимость в данных имеет вид $Y = f(X) + \varepsilon$, где
    * $\varepsilon$ - случайный шум,
    * $\mathbb{E} \varepsilon = 0$,
    * $\mathbb{D} \varepsilon = \sigma_{\varepsilon}^2 < +\infty$
2) Если новый объект $X_i$ случаный (у нас это так: $X_i \sim \mathcal{N}(0, \sigma^2)$)

Тогда:

$\text{MSE}(\hat{y}) = \mathbb{E}_{X}[(\mathbb{E}[\hat{y}(X) | X] - \mathbb{E}[Y|X])^2 + \mathbb{D}(\hat{y}(X) | X) + \mathbb{D}(Y | X)]$

Теперь от теории перейдем к нашей задаче:

$Y = X^2 + \varepsilon$, где $X$ - это один одномерный смлучаный объект, $\varepsilon$ - это одно из чисел $\varepsilon_i$

$\hat{y}(x) = \hat{\theta} * x$

1. Пусть $A(X) = (\mathbb{E}[\hat{y}(X) | X] - \mathbb{E}[Y|X])^2$
2. Пусть $B(X) = \mathbb{D}(\hat{y}(X) | X)$
3. Пусть $C(X) = \mathbb{D}(Y | X)$

Фиксируем $X = x$ - где $x$ это какое-то число, так как по условию объекты одномерны

Посчитаем $A(x), B(x), C(x)$

По свойству УМО (Лемма из теорвера), вместо везде где стоит $x$ малое, подставим $X$

от полученного возьмем ожидание $\mathbb{E}(x)$

1. Разбираемся с $A(x)$

* $\mathbb{E}[Y|X = x] = \mathbb{E}[X^2 + \varepsilon|X = x] = \mathbb{E}[x^2 + \varepsilon] = x^2 + \mathbb{E}[\varepsilon] = x^2 + 0 = x^2$
* $\mathbb{E}[\hat{y}(X) | X = x] = \mathbb{E}[\hat{y}(x)] = \mathbb{E}[\hat{\theta} \cdot x] = x \cdot \mathbb{E} \hat{\theta}$
* Итак, $A(x) = (\mathbb{E}[\hat{y}(X) | X = x] - \mathbb{E}[Y|X = x])^2 = (x^2 - x \cdot \mathbb{E} \hat{\theta})^2 = x^2 (x - \mathbb{E} \hat{\theta})^2$

Что такое $\mathbb{E} \hat{\theta}$? Мы можем сказать, потому что мы знаем, как выглядит $\hat{\theta}$ для МНК (см первую лекцию ML), $\hat{\theta}_{\text{МНК}} = (X^T X)^-1 \cdot X^T y$, только тут не большая коллизия обозначений но думаю понятно, что здесь имеется ввиду что X - это наша матрица объекты-признаки, а y - весь столбец таргетов для каждого объекта


2. Разбираемся с $B(x)$

$B(x) = \mathbb{D}(\hat{y}(X) | X = x) = \mathbb{D}(\hat{y}(x)) = \mathbb{D}(\hat{\theta} \cdot x) =$, так как $x$ константа фиксирована, то $= x^2 \cdot \mathbb{D} \hat{\theta}$, и опять же мы значем, что такое $\mathbb{D} \hat{\theta}$

3. Разбираемся с $C(x)$

$C(x) = \mathbb{D}(Y(X) | X = x) = \mathbb{D}(Y(x)) = \mathbb{D}(x^2 + \varepsilon) = \mathbb{D}(x^2) + \mathbb{D} \varepsilon_i$, поскольку данные иксы и шум эпсилон незавсиимы, значит дисперсия развалилась в сумму и к тому же мы рассматриваем для фиксированного объекта, так что $\mathbb{D} \varepsilon = \mathbb{D} \varepsilon_i$. Так как сечас $x^2$ - это фиксированная одномерная константа, то:

$C(x) = \mathbb{D} \varepsilon_i$

4. Итоговое разложение под ожиданием $\mathbb{E}_{X}$:

при фиксированном $X = x$:

$A(x) + B(x) + C(x) = x^2 (x - \mathbb{E} \hat{\theta})^2 + x^2 \cdot \mathbb{D} \hat{\theta} + \mathbb{D} \varepsilon_i$

начинаем варьировать один одномерный объект $X$:

$A(X) + B(X) + C(X) = X^2 (X - \mathbb{E} \hat{\theta})^2 + X^2 \cdot \mathbb{D} \hat{\theta} + \mathbb{D} \varepsilon_i$

5. от полученного берем ожидание $\mathbb{E}_{X}$

по свойству линейности нам необходимо и достаточно посчитать отдельно:

* $\mathbb{E}_{X}(A(X)) = \mathbb{E}_{X}[X^2 (X - \mathbb{E} \hat{\theta})^2]$

остается раскрыть скобки:

$X^2 (X - \mathbb{E} \hat{\theta})^2 = X^2 (X^2 - 2 \cdot X \cdot \mathbb{E} \hat{\theta} + (\mathbb{E} \hat{\theta})^2) = X^4 - 2 \cdot X^3 \cdot \mathbb{E} \hat{\theta} + X^2 \cdot (\mathbb{E} \hat{\theta})^2$

Мы знаем, что для нормального распределения $\mathcal{N}(0, \sigma^2)$ верно:

$\mathbb{E}_{X} X^2 = \sigma^2$, так как это просто дисперсия, из которой вычитается квадрат матожа обычного который равен нулю

$\mathbb{E}_{X} X^3 = 0$

$\mathbb{E}_{X} X^4 = 3 \sigma^4$

Тогда получаем, что:

* $\mathbb{E}_{X}(A(X)) = \mathbb{E}_{X}[X^4 - 2 \cdot X^3 \cdot \mathbb{E} \hat{\theta} + X^2 \cdot (\mathbb{E} \hat{\theta})^2] = 3 \sigma^4 + \sigma^2 \cdot (\mathbb{E} \hat{\theta})^2 = \text{bias}^2$

* $\mathbb{E}_{X}(B(X)) = \mathbb{E}_{X}(X^2 \cdot \mathbb{D} \hat{\theta}) = \mathbb{D} \hat{\theta} \cdot \mathbb{E}_{X}(X^2) = \mathbb{D} \hat{\theta} \cdot \sigma^2 = \text{variance}$, поскольку $\hat{\theta} = \hat{\theta}_{\text{МНК}}$ считается по обучающей выборке, а мы сейчас работаем с ново прибывшим объектом $X$, как из тестовой выборки, так что $\mathbb{D} \hat{\theta}$ выносится как константа из под матожа.

* $\mathbb{E}_{X}(C(X))$ = $\mathbb{E}_{X}(\mathbb{D} \varepsilon_i) = \mathbb{D} \varepsilon_i = \text{noise}$

Так как дисперсию отвечающую за разброс можно не доводить до конца, то я пожалуй оставлю $\mathbb{D} \hat{\theta}$

А вот видимо в смещении надо в явном виде расписать $\mathbb{E} \hat{\theta}$

Что такое $\mathbb{E} \hat{\theta}$? Мы можем сказать, потому что мы знаем, как выглядит $\hat{\theta}$ для МНК (см первую лекцию ML), $\hat{\theta}_{\text{МНК}} = (X^T X)^-1 \cdot X^T y$, только тут не большая коллизия обозначений но думаю понятно, что здесь имеется ввиду что X - это наша матрица объекты-признаки $X = (X_1, ..., X_n)^T$, а $y = (Y_1, ...., Y_n)^T$ - весь столбец таргетов для каждого объекта

$\hat{\theta}_{\text{МНК}} = (X^T X)^-1 \cdot X^T y = \frac{\sum_{i = 1}^n X_i \cdot Y_i}{\sum_{i = 1}^n X_i^2} = \frac{\sum_{i = 1}^n X_i \cdot (X_i^2 + \varepsilon_i)}{\sum_{i = 1}^n X_i^2} = \frac{\sum_{i = 1}^n X_i^3}{\sum_{i = 1}^n X_i^2} + \frac{\sum_{i = 1}^n X_i \varepsilon_i}{\sum_{i = 1}^n X_i^2}$, последнее слагаемое если поставить под матожидание в силу независимости шума и данных и линейности матожидания будет обнуляться, поскольку матож шума ноль. Значит:

Итак, $\mathbb{E} \hat{\theta} = \mathbb{E} \frac{\sum_{i = 1}^n X_i^3}{\sum_{i = 1}^n X_i^2}$ вроде больше это никак не упрощается.

Итак имеем:

$$\text{MSE}(\hat{y}) = \text{bias}^2 + \text{variance} + \text{noise} = [3\sigma^4 + \sigma^2 \cdot(\mathbb{E} [\frac{\sum_{i=1}^n X_i^3}{\sum_{i=1}^n X_i^2}])^2] + [\mathbb{D}\hat{\theta}\cdot\sigma^2] + [\mathbb{D}\varepsilon_i]$$

Теперь будем строить графики с помощью одного случайного леса. Первый график, который мы теперь построим - график зависимости метрики MSE от кол-ва деревьев в случайном лесу.

То есть мы один раз обучилии лес на ста деревьях, а теперь будем кумулятивно идти по деревьям от первого дерева до сотого и собирать у них значения MSE и усреднять на префиксе. И посмотрим куда по УЗБЧ эта штука сойдется

```python
from typing import Dict, Optional, Callable

def plot_some_lines(
    lines: Dict[str, np.ndarray],
    xlabel: str,
    estrimator_range: Optional[np.ndarray] = None,
    title: Optional[str] = None,
    metric_name: str = 'MSE',
):
    '''
    Отрисовывает линии из словаря lines

    Параметры:
    lines (Dict[str, np.ndarray]): словарь линий, ключ - label, значение - линия
    xlabel (str): название оси X
    estrimator_range (np.ndarray): значения на оси X
    metric_name (Optional[str]): имя метрики
    title (str): заголовок графика
    '''

    with sns.axes_style("darkgrid"):
        plt.figure(figsize=(12, 6))
        plt.title(title, fontsize=20)

        for name, line in lines.items():
            if estrimator_range is not None:
                plt.plot(estrimator_range, line, label=name, lw=3)
            else:
                plt.plot(line, label=name, lw=3)
        plt.grid(ls=':')
        plt.xlabel(xlabel)
        plt.ylabel(f'Метрика {metric_name}')
        plt.legend(fontsize=14)
        plt.show()


def cum_metric(
    model: RandomForestRegressor,
    metric: Callable[[np.ndarray, np.ndarray], float],
    x_test: np.ndarray,
    y_test: np.ndarray,
    shuffle: bool = False,
    binary_ths: float = 0.5,
):
    '''
    Считает значения метрики metric в зависимости от количества деревьев
    в обученной модели model метом композиции деревьев.

    Параметры:
    model (RandomForestRegressor): модель RandomForest
    metric (Callable[[np.ndarray, np.ndarray], float]): метрика
    x_test (np.ndarray): тестовая выборка
    y_test (np.ndarray): ответы на тестовой выборке
    shuffle (bool): надо ли перемешивать деревья
    binary_ths (float): порог вероятности для истинного класса
                       в случае задачи классификации

    Возвращает: кумулятивные значения метрики
    '''

    # Поведение для классификатора и регрессии будет немного отличаться
    is_classifier = isinstance(model, RandomForestClassifier)

    # Считаем предсказания по каждому дереву отдельно
    predictions_by_estimators = [
        est.predict_proba(x_test)[:, 1] if is_classifier else est.predict(x_test)
        for est in model.estimators_
    ]

    if shuffle:
        random.shuffle(predictions_by_estimators)

    numerator = np.cumsum(predictions_by_estimators, axis=0)
    n = len(predictions_by_estimators)
    denominator = (np.arange(n) + 1)[:, np.newaxis]
    cumpred = numerator / denominator

    if is_classifier:
        cumpred = cumpred > binary_ths

    cumacc = [metric(y_test, pred) for pred in cumpred]
    return np.array(cumacc)



mse_score = lambda y_true, y_pred: mean_squared_error(y_true, y_pred)

cum_mse_train = cum_metric(random_forest, mse_score, X_train.to_numpy(), y_train)
cum_mse_test = cum_metric(random_forest, mse_score, X_test.to_numpy(), y_test)
estrimator_range = np.arange(random_forest.n_estimators) + 1

plot_some_lines(
    lines={
        'Тренировочная выборка': cum_mse_train,
        'Тестовая выборка': cum_mse_test,
    },
    xlabel='n_estimators',
    estrimator_range=estrimator_range,
    title=f'Зависимость MSE от n_estimators',
    metric_name=f'MSE',
)
```

При 20 деревьев вся дисперсия ответов отдельно взятого дерева (неустойчивость одной модели) уже подавилась.

То есть в бэггинге (Bootstrap Aggregation), когда каждому дереву отдается выборка с потоворениями того же размера что и размер обучающей выборке, решается основная проблема деревьев как отдельно взятых сущностей - неусточивость модели дерева (если чуть-чуть поменять данные - предсказания дерева поменяются). За счёт чего так происходит? За счет того, что каждое из ста наших деревьев обучается на своей случайной подвыборке (с повторениями) элементов обучающих данных, отсюда берется случайность того, какие ответы выдают наши деревья и при усреднении получается неплохой результат (относительно неплохой, ведь в данной задаче MSE минимальное на тестовой выборке около 2.5 * 10^9 млн^2, значит ошибаемся мы примерно на 250000 долларов примерно). Однако важно отметить, что в данной задаче случайного выбора признаков не происходит, так как по умолчанию в sklearn для регрессии в каждом дереве берутся все признаки.

Теперь зафиксируем n_estimators равным 100 и построим график зависимости от max_depth на обучающей и на тестовой выборках.

```python
from typing import Sequence, Any
import copy

def get_train_and_test_mse(
    param_name: str,
    grid: Sequence[Any],
    other_params_dict: Dict[str, Any] = {},
):
    '''
    Функция для оценки точности классификации MSE
    для заданных значений параметра param_name

    Параметры:
    param_name (str): название параметра, который собираемся варьировать
    grid (Sequence[Any]): сетка значений параметра
    other_params_dict (Dict[str, Any]): словарь со значениями остальных параметров
    '''

    train_mse, test_mse = [], []
    params_dict = copy.copy(other_params_dict)

    for param_value in tqdm(grid):
        params_dict.update({param_name: param_value})
        estimator = RandomForestRegressor(**params_dict, random_state=RANDOM_STATE, n_jobs=-1)
        estimator.fit(X_train, y_train)

        train_mse.append(mean_squared_error(y_train, estimator.predict(X_train)))
        test_mse.append(mean_squared_error(y_test, estimator.predict(X_test)))

    return train_mse, test_mse


def plot_dependence(
    param_name: str,
    grid: Sequence[Any] = range(2, 20),
    other_params_dict: Dict[str, Any] = {}
):
    '''
    Функция для отображения графика зависимости MSE
    от значения параметра c названием param_name

    Параметры:
    param_name (str): название параметра, который собираемся варьировать
    grid (Sequence[Any]): сетка значений параметра
    other_params_dict (Dict[str, Any]): словарь со значениями остальных параметров
    '''

    train_mse, test_mse = get_train_and_test_mse(
        param_name, grid, other_params_dict
    )

    plot_some_lines(
        lines={
            'Тренировочная выборка': train_mse,
            'Тестовая выборка': test_mse,
        },
        xlabel=param_name,
        estrimator_range=grid,
        title=f'Зависимость MSE от {param_name}',
        metric_name=f'MSE',
    )
```

Теперь то же самое но на языке математики:

Пусть наши данные подчинены истинному закону $y = f(X, \varepsilon)$, а наши пердсказания это $\hat{y}$, тогда: запишем, то что доказали на лекции

$\text{MSE} \hat{y} = \mathbb{E}(\hat{y}(x) - \mathbb{E} f(x, \varepsilon))^2 + \mathbb{D} \hat{y}(x) + \mathbb{D} f(x, \varepsilon)$, где
* первый член это смещение модели
* второчой член это дисперсия ответов / предсказаний модели
* третий член - это дисперсия шума

То есть рассмотрим одно дерево в нашем лесу у которого огромная глубина:

оно гарантированно переобучилось, почему? При переобучении мы будем наблюдать, что у дерева большой MSE, из-за того что:

чем больше у дерева глубина, тем больше у него вершин, в которых очень много правил для классификации наших признаков, то есть смещение дерева (первый член) очень маленькое (при огромной глубине дерево "выучило" обучающую выборку). Но при этом если чуть-чуть поменять случайный шум в данных, то поменяются сами данные и поменяются предсказания модели, а именно х дисперсия может вырасти, впрочем как может подрасти и дисперсия самого шума.

Чтобы это исправить была придумана идея бэггинга, а именно построить $T$ штук независимых одинаково распределенных деревьев, где $T$ большое ($T = 100$ в нашей задаче). Важно так же отметить, что мы работаем с моделью где шум аддитивен (у него матож ноль, а данные наши подчиняются закону $y = f(X) + \varepsilon)$.

Тогда общее предсказание леса $\hat{y} = \frac{1}{T} \sum_{t = 1}^T \hat{y_t(x)}$, где $\hat{y_t}(x)$ - предсказание одного дерева из ансамбля леса.

Почему же нет переобучения? Потому что на лекции вывели формулу дисперсии предсказаний такого ансамбля / леса:

$\mathbb{D} \hat{y} = \mathbb{D} \hat{y_1}(x) \cdot (\frac{1}{T} + \frac{T - 1}{T} \cdot \text{corr}(\hat{y_1}(x), \hat{y_2}(x)))$

При маленькой корреляции между деревьями и при большом $T$ как раз дисперсия предсказаний леса становится меньше, что означает что модель становится устойчивой к шуму в таргете, случайным ошибкам в данных и прочее. Как раз поэтому и нет переобучения.

$\mathbb{D} \hat{y} = \mathbb{D} \hat{y_1}(x) \cdot (\frac{1}{T} + \frac{T - 1}{T} \cdot \text{corr}(\hat{y_1}(x), \hat{y_2}(x)))$

При маленьких значения max_fetures деревья не имеют выбора, им дали по один-два признака и они с ними работают. С одной стороны, это хорошо, потому что деревья будут иметь низкую коррелированность, поскольку работаю с разными признаками, но с другой стороны:, такие деревья обречены отдкльно давать очень смещенные предсказания (далекие от истинного значения таргета) и с большой дисперсией предсказаний (не робастны к шуму), что означает, что дисперсия предсказаний всего леса будет высокой, так как в ней стоит множитель $\mathbb{D} \hat{y_1}(x)$.

При больших значениях max_fetures деревья работаб с почти всеми признаками за счет чего получаются сильно скоррелированными с одной стороны, но с другой стороны, они могут классно отдельть один объект от другого на обучающей выборке так точно, за счет чего маленький множитель дисперсии $\mathbb{D} \hat{y_1}(x)$ и большое $T$, дадут лучшее предсказание всего леса.

Собственно, на графике мы именно это и видим. При увеличении max_features (начиная примерно с 3) MSE начинает приближаться к своему оптимальному значению, то есть выходить на плато.

```python
def get_train_and_test_mse(
    param_name: str,
    grid: Sequence[Any]
):
    '''
    Функция для оценки точности классификации MSE
    для заданных значений параметра param_name

    Параметры:
    param_name (str): название параметра, который собираемся варьировать
    grid (Sequence[Any]): сетка значений параметра
    '''

    train_mse, test_mse = [], []
    params_dict = {}

    for param_value in tqdm(grid):
        params_dict.update({param_name: param_value})
        estimator = BaggingRegressor(
                        estimator=DecisionTreeRegressor(random_state=RANDOM_STATE),
                        **params_dict, # max_features
                        n_estimators=100, # по умолчанию
                        n_jobs=-1
                    )

           # было так RandomForestRegressor(**params_dict, random_state=RANDOM_STATE, n_jobs=-1)
        estimator.fit(X_train, y_train)

        train_mse.append(mean_squared_error(y_train, estimator.predict(X_train)))
        test_mse.append(mean_squared_error(y_test, estimator.predict(X_test)))

    return train_mse, test_mse


def plot_dependence(
    param_name: str,
    grid: Sequence[Any] = range(2, 20)
):
    '''
    Функция для отображения графика зависимости MSE
    от значения параметра c названием param_name

    Параметры:
    param_name (str): название параметра, который собираемся варьировать
    grid (Sequence[Any]): сетка значений параметра
    '''

    train_mse, test_mse = get_train_and_test_mse(
        param_name, grid
    )

    plot_some_lines(
        lines={
            'Тренировочная выборка': train_mse,
            'Тестовая выборка': test_mse,
        },
        xlabel=param_name,
        estrimator_range=grid,
        title=f'Зависимость MSE от {param_name}',
        metric_name=f'MSE',
    )


plot_dependence("max_features", list(range(1, 8 + 1)))
```

**Сравним как работают две модели BaggingRegressor и RandomForest**

Bagging Regressor и RandomForest одинаково работают с данными, то есть берется бустрапированная выборка для каждого дерева из ансамбля (выборка с повторениями из обучающей выборки того же размера что обучающей выборки), об этом мы говорили выше.

1. Формула декомпозиции ошибки MSE

Пусть наши данные подчинены истинному закону $y = f(X, \varepsilon)$, а наши пердсказания это $\hat{y}$, тогда: запишем, то что доказали на лекции

$\text{MSE} \hat{y} = \mathbb{E}(\hat{y}(x) - \mathbb{E} f(x, \varepsilon))^2 + \mathbb{D} \hat{y}(x) + \mathbb{D} f(x, \varepsilon)$, где
* первый член это смещение модели
* второчой член это дисперсия ответов / предсказаний модели
* третий член - это дисперсия шума

2. Формула дисперсии ансамбля

Тогда общее предсказание леса $\hat{y} = \frac{1}{T} \sum_{t = 1}^T \hat{y_t(x)}$, где $\hat{y_t}(x)$ - предсказание одного дерева из ансамбля леса.

На лекции вывели формулу дисперсии предсказаний такого ансамбля / леса:

$\mathbb{D} \hat{y} = \mathbb{D} \hat{y_1}(x) \cdot (\frac{1}{T} + \frac{T - 1}{T} \cdot \text{corr}(\hat{y_1}(x), \hat{y_2}(x)))$

3. Картинка из интернета про смысл обучения и недообучения



**Наша цель** - это построить идеальную модель, которая должна удовлетворять критериям:

   * У нее должны быть "классные" предсказания, которые в среднем неплохо приближают истинную зависимость таргета от данных.
   * Она должна быть устойчивой, то есть при небольшом изменении данных, дисперсия ее ответов должна быть стабильной, то есть ответы сильно меняться не должны (это свойство так же можно трактовать как способность модели обобщать)

**Проблемы отдельно взятого дерева:**

Оно может сильно переобучаться, то есть смещение будет маленькое на трэйн выборке, но при этом дисперсия большая (нет обощающей способности на тесте).

И RandomForest и BoostrapRegresso с этим борются, поскольку усредняют ответы многих независимых деревьев. Но какое из них лучше станет ясно, когда мы сравним наши их по двум критериям выше

4. Что со смещением?

**Bagging Regressor** рандомно выбирает max_features признаков, что в каждой вершине дерево выбирает по какому из данных признаков ему разбивать вершину на две. А что если попали неважные признаки в эти max_features признаков для одного дерева? Тогда дерево будет плохо предсказывать наши данные, и даже увеличение глубины его не спасет, он обучился на признаках которые никакого вклада в r2_score для объяснения таргета не вносят. А раз так, то у этого дерева смещение будет больше. Что означает, что даже после того, как мы просуммируем и усредним предсказания (независимые и одинаково распределенные случайные величины) всех деревьев, смещение итоговой модели (ансамбля) подрастет, поскольку на лекции из незавсимости и одинаковой распределенности, мы поняли, что смещение ансамбля совпадает со смещением каждого отдельновзятого дерева.

**Random Forest** рандомно выбирает признаки не для одного конкретного дерева. Он берет одно дерево и в каждой вершине происходит рандомный выбор признаков в количестве max_features штук. То есть это свойство круче, чем у Bagging Regressor. Если мы вспомним работу одного дерева, то можно сказать, что оно работает жадно. С геометрической точки зрения, в каждой вершине мы выбираем наиболее важное направление (признак). Допустим в родителе мы зафиксировали какой-то признак. Какова вероятность того, что этот же признак мы выберем в одном из детей? Она снижается, за счет того, что дерево строится так что минимизируется мера хаоса в листьях, то есть как раз выстраивается способность обощать. Значит в детях мы будем скорее всего выбирать "ортогональные" направления (а именно другие признаки). Здесь очень хорошо привести аналогию с PCA. За счет такого подхода каждое дерево при оптимальной глубине имеет небольшое смещение (выбросы как в BaggingRegressor не влияют). А значит и весь ансамбль имеет такое же смещение - небольшое.

5. Что с дисперсией?

**Bagging Regressor** он сгенерировал пачку индексов один раз и на все дерево. Те деревья, которые получили похожие пачки индексов, будут строится практически идентично, следуя важности признака жадно. За счет этого корреляция между деревьями высокая. При этом дисперсия одного дерева, в которое попали не самые важные признаки, будет немаленькой, потому что оно выучило не самые релевантные закономерности. Следовательно его способность обобщать данные будет не очень хорошей, потому что оно оперирует не большинством признаков внутри всего дерева как Random Forest, а все дерево умеет оперировать скажем только 3 мя признаками, ясно, что обобщательная способность там никакая, а это в точности есть дисперсия ответов - слепое дерево получилось. А раз так, то $\mathbb{D} \hat{y_1}(x)$ и $\text{corr}(\hat{y_1}(x), \hat{y_2}(x))$ в формуле дисперсии ансамбля выше большие, значит дисперсия ансамбля большая

**Random Forest** за счет того, что в каждой вершине происходит что-то вроде PCA с элементами рандома (то есть в каждой вершине рандомна выбирается новая пачка признаков рандомно), то есть в каждой вершине есть очень много вариантов, какие индексы признаков туда попадут. Но всегда признак выбирается самый важный из сэмплированных и на следующих уровня вероятность выбрать тот же признак маленькая. Это делает деревья более независимыми, то есть их корреляции становится маленькой. Если взглянуть на формулу дисперсии ансамбля в пункте 2, то тсновится ясно, что $\frac{1}{T} \to 0$, и $\frac{T}{T - 1} \cdot \text{corr}(\hat{y_1}(x), \hat{y_2}(x))$ становится маленьким, то вся дисперсия ансамбля становится меньше.

Насколько уменьшается разброс в каждом случае? Для каждого случая постройте также матрицу корреляций предсказаний базовых моделей и гистограмму по ним.

```python
bagging_trees = BaggingRegressor(
                        estimator=DecisionTreeRegressor(max_features=6, max_depth=21, random_state=RANDOM_STATE), # чтоб как Random Forests
                        n_estimators=100,
                        n_jobs=N_JOBS,
                        random_state=RANDOM_STATE
                    )

trees_pipeline = Pipeline(steps=[
    ('classifier', bagging_trees)])


from sklearn.linear_model import Ridge
import warnings
warnings.filterwarnings(
    "ignore",
    message="X has feature names, but .* was fitted without feature names",
    category=UserWarning,
)
# UserWarning: X has feature names, but Ridge was fitted without feature names
# warnings.warn(

bagging_regression = BaggingRegressor(
                        estimator=Ridge(),
                        n_estimators=100,
                        n_jobs=N_JOBS,
                        random_state=RANDOM_STATE
                    )
# прежде чем на таких данных делать лин регрессию, надр сделать Scaling
preprocessor = ColumnTransformer(
    transformers=[
        ('transform_all_columns',
         StandardScaler(),                              # что применяем
         houses_X.columns),                             # к чему применяем

    ]
)
ridge_pipeline = Pipeline(steps=[
    ('preprocessor', preprocessor),
    ('classifier', bagging_regression)])


def analyze_ensemble_correlations(ensemble_pipeline, X_test, model_name="Ансамбль"):
    print(f"=== Анализ для: {model_name} ===")
    ensemble_pipeline.fit(X_train, y_train)
    y_pred_ensemble = ensemble_pipeline.predict(X_test)
    print(f"Дисперсия бэггинга: {np.var(y_pred_ensemble):.4f}")

    preds = np.array([est.predict(X_test) for est in ensemble_pipeline['classifier'].estimators_])
    corr_matrix = np.corrcoef(preds)
    corr_mean = np.mean(corr_matrix)
    print(f"Средняя корреляция: {corr_mean:.4f}")

    print(f"Корреляция для двух базовых моделей: {corr_matrix[0, 1]}")

    ests_cnt = len(ensemble_pipeline['classifier'].estimators_)
    theor_ratio = 1/ests_cnt + (ests_cnt - 1) / ests_cnt * corr_matrix[0, 1]
    print(f"Отношение дисперсий D(бэггинг_preds) / D(базовая_preds)) какое должно быть в теории: {theor_ratio:.4f}")

    practical_ratio = np.var(y_pred_ensemble) / np.var(preds[0])
    print(f"Отношение дисперсий D(бэггинг_preds) / D(базовая_preds)) на практике: {practical_ratio:.4f}")


    fig, axes = plt.subplots(1, 2, figsize=(15, 6))

    sns.heatmap(corr_matrix, cmap="coolwarm", ax=axes[0])
    axes[0].set_title(f"Матрица корреляций предсказаний базовых моделей для модели - {model_name}")
    axes[0].set_xlabel("Индекс базовой модели")
    axes[0].set_ylabel("Индекс базовой модели")


    sns.histplot(corr_matrix.flatten(), kde=True, color='blue', ax=axes[1])
    axes[1].axvline(corr_mean, color='red', linestyle='--', label=f'Среднея corr= {corr_mean:.3f}')
    axes[1].set_title(f"Гистограмма распределения значений корреляции\n({model_name})")
    axes[1].set_xlabel("Коэффициент корреляции")
    axes[1].set_ylabel("Частота")
    axes[1].legend()

    plt.tight_layout()
    plt.show()

analyze_ensemble_correlations(trees_pipeline, X_test, model_name="Лес")
analyze_ensemble_correlations(ridge_pipeline, X_test, model_name="Ансамбль Ridge Регрессий")
```

В этой задаче вам предлагается расмотреть и сравнить две стратегии поиска оптимальных значений некоторого функционала: **Grid Search** и **Random Search**. В реальной жизни значение функционала &mdash; это значение целевой метрики качества, а оптимизация идет по гиперпараметрам некоторой модели машинного обучения. Коротко напомним, в чем же заключается каждая стратегия.

* **Grid Search** &mdash; полный перебор по сетке. Суть перебора по сетке заключается в следующем: вы задаете область значений гиперпараметров, которые хотите оптимизировать, обучаете все модели, которые возможно получить в рамках заданного пространства поиска гиперпараметров. Затем выбираете из данных моделей ту, которая дала наилучшее качество на кросс-валидации.
* **Random Search** &mdash; это стратегия, при которой рассматривается некоторая выборка из множества значений гиперпараметров. По умолчанию часто используется равномерное распределение. Лучшая модель выбирается так же на основе кросс-валидации.

Разберем сначала **одномерный случай**. Для этого рассмотрим следующую функцию.

```python
def f(x):
    return 5.2 - 5/(1+(3*(x))**2) - np.cos(x)/5
```

Поcтройте график функции $f(x)$ на отрезке $[-20, 20]$. В чем его особенность?

```python
x_grid = np.linspace(-20, 20, 10000)
y_grid_values = f(x_grid)
plt.figure(figsize=(12, 7))
plt.plot(x_grid, y_grid_values)
plt.show()
```

Подумайте, какие проблемы могут возникнуть при поиске минимума этой функции, если используется поиск по сетке?

**Ответ:**

Рассмотрим что было на лекции:

На графике мы видим функцию с пиком



Можно видеть что при поиске минимума этой функции по сетке будет задействовано недостаточное число точек, поскольку они идут с некоторым шагом, они запросто могут проскочить глобальный минимум и тогда скорее всего минимум который находится в пике найден не будет а будет надйен один из локальных периодических минимумов



А при поиске рандомном будет сгенерировано достаточно для того, чтобы найти истинный миниммум функции

Рассмотрим **поиск по сетке** и проведем вычислительный эксперимент. Для этого будем генерировать случайное смещение из $U[-0.5, 0.5]$, а затем рассматривать смещенную на $x_0$ сетку размера $n$ на отрезке $[-10, 10]$, после чего поиском по сетке искать оптимальное значение $f$.

```python
# рассматриваемые размеры сеток
grid_size = np.arange(2, 1000)
n_experiments = 100  # количество экспериментов
```

```python
# массив для минимумов по сетке для текущего эксперимента
grid_min_f_trace = []

for _ in tqdm(range(n_experiments)):
    # случайное смещение сетки на [-0.5, 0.5]
    x_0 = np.random.uniform(-0.5, 0.5)
    # массив для минимумов для текущего размера сетки
    grid_f_min = []
    # цикл по размеру сетки
    for n in grid_size:
        # сетка на [-10, 10] размера n
        grid = np.linspace(-10, 10, n) + x_0
        # найденный по сетке со смещением x_0 минимум функции f
        f_min = np.min(f(grid))
        # сохраним минимум для сетки размера n
        grid_f_min.append(f_min)
    # сохраним минимумы текущего эксперимента
    grid_min_f_trace.append(grid_f_min)
```

Для нескольких экспериментов постройте графики значений минимума от количества точек в сетке.
Для большей наглядности и информативности сразу **зафиксируйте диапазон значений по оси y**, а также отображайте графики в **логарифмическом масштабе**.

```python
# Для первого эксперимента
plt.figure(figsize=(10, 7))

plt.plot(grid_size, grid_min_f_trace[0], label="Минимум, к-рый нашел GridSearch (Эксперимент №1)")
plt.plot(grid_size, grid_min_f_trace[-1], label="Минимум, к-рый нашел GridSearch (Эксперимент №100)")
plt.axhline(y=f(0), color="red", label="Истинное значение минимума") # минимум в нуле

plt.title("Зависимость значений минимума, найденных GridSearch'ем, от кол-ва точек в сетке")
plt.xlabel("Кол-во точек в сетке")
plt.ylabel("Значения минимума")
plt.ylim(10**(-17), 10)
plt.yscale('log')
plt.legend()
plt.show()
```

Вывод: поиск по сетке не справился найти реальный миниммум, как и ожидалось

Немного про то, почему линии на графике выглядят так:

наша функция имеет в себе косинус, то есть функция наша является периодической. Когда точек в сетке мало, то шаг в np.linspace(-10, 10, n) для сетки будет большим, поэтому мы часть будем оказываться в локальных мимнимумах (к-рые есть у функции из-за косинуса). По мере возрастания числа точек в сетке, шаг (расстояние между двумя соседними точками, рассматриваемыми Grid Search) будет становится меньше. А раз так, то у нас больше шансов попасть в как можно нижнее значение локального минимума функции. За счёт этого, обе линии с ростом кол-ва точек убывают.

Первый эксперимент - зеленая линия, на нем пики идут примерно два раза чаще чем на оранжевой линии (100ый эксперимент). Скорее всего, это связано со случаныйм смещением $x_0$. В первом эксперименте оно было таким, что шаги сетки могли попасть к глобальному минимуму (то есть очень близко к нему находились), то есть скорее всего шаги сетки попадали на склоны этого глобального минимума, и поэтому значения минимума очень колеблются. А в 100м эксперименте, скорее всего шаги сетки просто ходили по холмикам которые образовалаись из-за косинуса и так близко к глобальному минимму не подбирались.

Теперь рассмотрим **случайный поиск**. Снова проведем вычислительный эксперимент. Обратите внимание, что случайные точки для эксперимента мы генерируем один раз.

```python
# рассматриваемое количество случайных точек
random_size = np.arange(2, 1000)
```

```python
# массив для минимумов для текущего эксперимента
random_min_f_trace = []

# цикл по экспериментам
for _ in tqdm(range(n_experiments)):
    # случайное смещение на [-0.5, 0.5]
    x_0 = np.random.uniform(-0.5, 0.5)
    # массив для минимумов для текущего количества точек
    random_min_f = []
    # сгенерируем случайные на [-10, 10] точки сразу
    points = np.random.uniform(-10, 10, size=1000)

    #цикл по количеству точек
    for n in random_size:
        # найденный по первым n случайным точкам со смещением x_0 минимум функции f
        f_min = np.min(f(points[:n] + x_0))
        # сохранение результатов
        random_min_f.append(f_min)

    random_min_f_trace.append(random_min_f)
```

Снова для нескольких экспериментов постройте графики значений минимума от количества точек при случайном поиске.

```python
plt.figure(figsize=(10, 7))

plt.plot(random_size, random_min_f_trace[0], label="Минимум, к-рый нашел RandomSearch (Эксперимент №1)")
plt.plot(random_size, random_min_f_trace[-1], label="Минимум, к-рый нашел RandomSearch (Эксперимент №100)")
plt.axhline(y=f(0), color="red", label="Истинное значение минимума") # минимум в нуле

plt.title("Зависимость значений минимума, найденных RandomSearch'ем, от кол-ва точек при случ поиске")
plt.xlabel("Кол-во точек при случ поиске")
plt.ylabel("Значения минимума")
plt.ylim(10**(-17), 10)
plt.yscale('log')
plt.legend()
plt.show()
```

Вывод по графику:

Видно, что даже Random Search в этой одномерной задаче оптимизации не справился дойти до глобального минимума. Но интересно, посмотреть на то, как устроен график:

Random Search считает минимум функции на первых n из изначально (до эксперимента) сгенерированных точек. За счет этого, график имеет такую ступенчатую форму. Допустим, к нам пришла точка $n + 1$ из набора изначально (до эксперимента) сгенерированных точек. Если значение функции в ней меньше, чем во всех точках до этого (от №1 до №n точки), то функция опускается вниз (переходя на новую полочку / плато). Если же значение функции в ней больше либа равно, чем во всех точках до этого (от №1 до №n точки), то поскольку Random Search считает минимум, то в данном случае минимум не обновляется (остаемся на полочке / плато)

Сравним два метода. Для этого постройте график зависимости найденного минимума от количества точек (случайных или в сетке). Перед этим предварительно усредните результат по количеству экспериментов. Какие результаты у вас получились?

```python
mean_grid_search = np.mean(grid_min_f_trace, axis=0)
mean_random_search = np.mean(random_min_f_trace, axis=0)

plt.figure(figsize=(10, 7))

plt.plot(grid_size, mean_grid_search, label="Grid Search")
plt.plot(grid_size, mean_random_search, label="Случайный поиск (Random Search)")
plt.axhline(y=f(0), color="red", label="Истинное значение минимума") # минимум в нуле

plt.title("Сравнение Grid Search и Random Search (усреднено по всем экспериментам)")
plt.xlabel("Количество точек")
plt.ylabel("Значения минимума")
plt.ylim(10**(-17), 10)
plt.yscale('log')
plt.legend()
plt.show()
```

Вывод по сравнительному графику:

Получается, что GridSearch при увеличении количества точек, в среднем по всем эксперимента, выдал результат лучше (значение минимумов меньше), чем RandomSearch. Но это было ожидаемо для подбора одного гиперпараметра - точки минимума x = 0 (одномерная задача оптимизации). При большом количестве точек в сетке получается очень маленький шаг, и получается что проверяется очень много точек, поэтому шансы найти реальную точку минимума высоки. При это random search будет работать лучше при многомерной оптимизации. В данном эксперименте RandomSearch случайно генерировал точки, поэтому какие-то "важные" точки оно могло не покрыть.

На графике все так же прослеживается "колеблющаяся" природа графика GridSearch (как он скачет вокруг глобального минимума, и не может к нему подобраться, слишком уж там крутые овраги, и даже начальное случайное смещение и увеличением кол-ва точек с вытекаюим уменьшением шага не способствует поимке глоабльного пикового минимума). А так же видна ступенчатая природа графика для Random Search, которая связана со свойствами поиска минимумам на префиксе как мы обсуждали выше.

Теперь расмотрим **двумерный случай**, для этого определим следующую функцию:

```python
def f(x, y,):
    return 5.4 - 5/(1+(3*x)**2) - np.cos(x)/5 - np.cos(y)/5
```

Как видно из картинки в десмосе, минимум функции достигается в точке (0, 0):



Постройте тепловую карту значений этой функции на $[-20, 20]^2$.

```python
x_grid = np.linspace(-20, 20, 1000)
y_grid = np.linspace(-20, 20, 1000)

X, Y = np.meshgrid(x_grid, y_grid)
Z = f(X, Y)

plt.figure(figsize=(10, 7))

contours = plt.contourf(X, Y, Z, levels=30, cmap="viridis")
cbar = plt.colorbar(contours)
cbar.set_label("f(x, y)", fontsize=12)

plt.title(r"Тепловая карта значений функции f(x, y) на $[-20, 20]^2$")
plt.xlabel("X")
plt.ylabel("Y")
plt.xlim(-20, 20)
plt.ylim(-20, 20)
plt.show()
```

Проведите эксперимент, аналогичный одномерному случаю. В случае поиска по сетке рассмотрите сетку размера $n$ на $[-10, 10]$ по каждой координате. Таким образом, общий размер сетки будет $n^2$. Для ускорения вычислений в случае случайного поиска будет достаточно рассмотреть только такое количество точек, которое соответствует рассмотренным размерам сетки (т.е. квадраты чисел).

Сравните результат с полученным ранее. Сделайте вывод.

1. Эскперименты с поиском минимума по сетке

```python
# рассматриваемые размеры сеток
grid_size = np.arange(2, 100 + 1)
n_experiments = 100  # количество экспериментов

# массив для минимумов по сетке для текущего эксперимента
grid_min_f_trace = []

for _ in tqdm(range(n_experiments)):
    # случайное смещение сетки на [-0.5, 0.5]
    x_0 = np.random.uniform(-0.5, 0.5)
    y_0 = np.random.uniform(-0.5, 0.5)
    # массив для минимумов для текущего размера сетки
    grid_f_min = []
    # цикл по размеру сетки
    for n in grid_size:
        # сетка на [-10, 10] размера n
        x_grid = np.linspace(-10, 10, n) + x_0
        y_grid = np.linspace(-10, 10, n) + y_0
        X, Y = np.meshgrid(x_grid, y_grid) # чтобы все пары точек перебрать
        # найденный по сетке со смещением x_0 минимум функции f
        f_min = np.min(f(X, Y))
        # сохраним минимум для сетки размера n
        grid_f_min.append(f_min)
    # сохраним минимумы текущего эксперимента
    grid_min_f_trace.append(grid_f_min)
```

График для GridSearch

```python
# Для первого эксперимента
plt.figure(figsize=(10, 7))

plt.plot(grid_size, grid_min_f_trace[0], label="Минимум, к-рый нашел GridSearch (Эксперимент №1)")
plt.plot(grid_size, grid_min_f_trace[-1], label="Минимум, к-рый нашел GridSearch (Эксперимент №100)")
plt.axhline(y=f(0, 0), color="red", label="Истинное значение минимума") # минимум в нуле

plt.title("Зависимость значений минимума, найденных GridSearch'ем, от кол-ва точек в сетке")
plt.xlabel("Кол-во точек в сетке по одной координате")
plt.ylabel("Значения минимума")
plt.ylim(10**(-17), 10)
plt.yscale('log')
plt.legend()
plt.show()
```

Здесь происходит примерно та же самая картина, что и в одномерной ситуации. Мы видим сильные колебания / гормошки. Это всё так же связаны с гармонической природой нашей двумерной функции (в ней все еще содержит косинус но теперь аж целых два по x, по y). Здесь значения минимума соседний не такие "дерганные" как было в одномерной ситуации. Я предполагаю, что это связано с тем, что у нас перебор точек ведется по парам (x, y). То есть если поменять какую-то одну из координат сильно результат не изменится.

При этом видно, что Эксперимент№100 (оранжевая линия) умудрился найти значение минимума меньше чем Эксперимент №1 (зеленая линия). Опять же это связано с выбором случайного смещения $(x_0, y_0)$. Где-то (Эксперимент№100 ) повезло так сместиться, что при достаточно большом кол-ве точек, где шаг маленький попасть на один из склонов, который ведет к пиковому минимуму, а где-то не повезло.

2. Эксперименты с поиском минимума рандомно

```python
# рассматриваемое количество случайных точек
random_size = np.arange(2, 100 + 1)**2

# массив для минимумов для текущего эксперимента
random_min_f_trace = []

# цикл по экспериментам
for _ in tqdm(range(n_experiments)):
    # случайное смещение на [-0.5, 0.5]
    x_0 = np.random.uniform(-0.5, 0.5)
    y_0 = np.random.uniform(-0.5, 0.5)
    # массив для минимумов для текущего количества точек
    random_min_f = []
    # сгенерируем случайные на [-10, 10] точки сразу
    points = np.random.uniform(-10, 10, size=(1000, 2))
    X, Y = points[:, 0], points[:, 1]
    #цикл по количеству точек
    for n in random_size:
        # найденный по первым n случайным точкам со смещением x_0 минимум функции f
        f_min = np.min(f(X[:n] + x_0, Y[:n] + y_0))
        # сохранение результатов
        random_min_f.append(f_min)

    random_min_f_trace.append(random_min_f)
```

```python
plt.figure(figsize=(10, 7))

plt.plot(random_size, random_min_f_trace[0], label="Минимум, к-рый нашел RandomSearch (Эксперимент №1)")
plt.plot(random_size, random_min_f_trace[-1], label="Минимум, к-рый нашел RandomSearch (Эксперимент №100)")
plt.axhline(y=f(0, 0), color="red", label="Истинное значение минимума") # минимум в нуле

plt.title("Зависимость значений минимума, найденных RandomSearch'ем, от кол-ва точек при случ поиске")
plt.xlabel("Кол-во точек при случ поиске")
plt.ylabel("Значения минимума")
plt.ylim(10**(-17), 10)
plt.yscale('log')
plt.legend()
plt.show()
```

Вывод по графику:

Природа графика все такая же, как была и в одномерном случае, поскольку RandomSearch все так же считает минимумы на перфиксе, просто теперь на двумерном префиксе. И из-за свойств минимума, обговоренных ранее, график имеет такую ступенчатую природу.

Единственное, что я могу отметить, это то, что по сравнению с одномерным случаем, здесь меньше полочек / ступенек в начале графика (при маленьком кол-ве точек). То есть как только RandomSearch нашел минимум, он продолжает постонно находить его же значение, стабилизировался, даже не смотря на то, что для каждого эксперимента по Методу Монте Карло сэмплируется разный набор точек. Видно, что в двумерном случае, двумерное заполнение плоскости рандомными точками все еще достаточно для удержания найденного минимума.

Сравним два метода. Для этого построим график зависимости найденного минимума от количества точек (случайных или в сетке). Перед этим предварительно усредним результат по количеству экспериментов. Какие результаты получились?


```python
mean_grid_search = np.mean(grid_min_f_trace, axis=0)
mean_random_search = np.mean(random_min_f_trace, axis=0)

plt.figure(figsize=(10, 7))

plt.plot(grid_size, mean_grid_search, label="Grid Search")
plt.plot(grid_size, mean_random_search, label="Random Search")
plt.axhline(y=f(0, 0), color="red", label="Истинное значение минимума") # минимум в нуле

plt.title("Сравнение Grid Search и Random Search (усреднено по всем экспериментам)")
plt.xlabel("Количество точек")
plt.ylabel("Значения минимума")
plt.ylim(10**(-17), 10)
plt.yscale('log')
plt.legend()
plt.show()
```

В двумерном случае Random Search показал себя лучше чем Grid Search, поскольку линия Random Sarch проходит над линией Grid Search. В конце концов, Random Search вышел на плато, а Grid Search продолжал нестабильно колебаться.

**Подведем итоги:**

* В одномерном случае, явное превосходством было за Grid Search, особенно при увеличении кол-ва точек в сетке, то есть при уменьшении шага между соседними точками. Поскольку Random Search генерировал точки случайно могли быть какие-то области под графиком функции f никак не затронутые Radnom Search'ем, то есть там значение функции не проверялась и из-а этого мы могли потерять глобальный миниму.
* В двумерном случае ситуации кардинально меняет. Random Search демонстрирует лучший результат. Почему так? При тех же самый "затратах", вычислительных мощностях вот что происходит:

Grid Search имеет сетку из 10 тыс точек. То есть по каждой координате проверяется 100 точек, соответсвенно шаг между точками в рамках однй координаты большой. Из-за этого вероятность промазать и не найти глобальный миниммум высока.

В то время как Random Search работает с 10 тысячими пар (x, y). То есть по каждой координате рассматривается уже не 100, а 10 тысяч точек. Конечно, у Random Search есть все шансы покрыть большую часть плоскости и найти глоабльный минимум. В данном случае даже Random Search не справился найти пиковый минимум (это была слишком сложная задача), но на многомерных задачах такого рода (с пиками) потенциал Random Search значительно выше, нежели чем у Grid Search, который пригоден для таких "сложных" (пиковых) задач только в одномерном случае

Почему без приставки Shuffle? В ShuffleSplit разбиение каждой итерации не зависит от предыдущих итераций, объект выборки может попасть как несколько раз в тестовые фолды, так и вовсе не попасть ни разу. Поэтому мне кажется, что KFold использовать в этом смысле чуть надежнее.

А вот что касается стратегии Group - у нас в данных очень много локаций (вот по ним можно сгруппировать)

**1.** Используя одну из стратегий кросс-валидации подберите оптимальные параметры обучения случайного леса &mdash; `n_estimators`, `max_depth` и `min_samples_split`. В качестве метрик используйте Accuracy и F1.

Делаем данные X, Y

```python
X = data.drop(columns=["RainTomorrow"])
y = data["RainTomorrow"]

X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.2,
    random_state=RANDOM_STATE,
    stratify=y # у нас в данных дисбаланс
)
```

```python
from sklearn.model_selection import StratifiedGroupKFold, GridSearchCV

stratified_group_kfold = StratifiedGroupKFold(
    n_splits=5,          # Количество фолдов
    shuffle=False,        # Перемешиваем ли данные перед разбиением
    # random_state=RANDOM_STATE    # Фиксируем рандом, который появляется из-за shuffle
)

model = RandomForestClassifier(random_state=RANDOM_STATE)

# Задаем сетку гиперпараметров для подбора
param_grid = {
    'n_estimators': [5, 10, 15, 20, 25, 30, 35, 40, 45, 50],
    'max_depth': [2, 5, 8, 12, 15],
    'min_samples_split': list(range(2, 8))
}

# Настраиваем GridSearchCV
grid_search = GridSearchCV(
    estimator=model,
    param_grid=param_grid,
    scoring=['accuracy', 'f1_macro'],       # Считаем обе метрики
    refit='f1_macro',                       # Финальную модель выбираем по качеству f1-score
    cv=stratified_group_kfold,
    n_jobs=-1,                        # -1 задействует все доступные ядра процессора для параллелизации
    return_train_score=False
)

# Запускаем поиск (ВАЖНО: передаем groups=X["Location"])
X_train_dropped = X_train.drop(columns=['Location']) if 'Location' in X_train.columns else X_train
X_train_encoded = pd.get_dummies(X_train_dropped, drop_first=True) # Это One Hot Encoding, в одном из материалов (семинаре вроде) делалось именно так (чтобы дозакодировать оставшиеся категориальные признаки, ибо дерево только с числами работает)

X_test_dropped = X_test.drop(columns=['Location']) if 'Location' in X_test.columns else X_test
X_test_encoded  = pd.get_dummies(X_test_dropped,  drop_first=True)

grid_search.fit(
    X=X_train_encoded,
    y=y_train,
    groups=X_train["Location"].values             # Передаем группы для StratifiedGroupKFold
)

# Выводим лучшие результаты
print(f"Лучшие параметры: {grid_search.best_params_}")
print(f"Лучший F1-score на кросс-валидации: {grid_search.best_score_:.4f}")

# Индекс лучшей комбинации в общей таблице результатов
best_index = grid_search.best_index_
best_accuracy = grid_search.cv_results_['mean_test_accuracy'][best_index]
print(f"Accuracy для этой же лучшей модели: {best_accuracy:.4f}")
```

```python
from sklearn.metrics import f1_score

def get_train_and_test_mse(
    param_name: str,
    grid: Sequence[Any],
    other_params_dict: Dict[str, Any] = {},
):
    '''
    Функция для оценки точности классификации
    для заданных значений параметра param_name

    Параметры:
    param_name (str): название параметра, который собираемся варьировать
    grid (Sequence[Any]): сетка значений параметра
    other_params_dict (Dict[str, Any]): словарь со значениями остальных параметров
    '''

    params_dict = copy.copy(other_params_dict)
    train_acc, test_acc, train_f1, test_f1 = [], [], [], []

    for param_value in tqdm(grid):
        params_dict.update({param_name: param_value})
        estimator = RandomForestClassifier(**params_dict, random_state=RANDOM_STATE, n_jobs=-1)
        estimator.fit(X_train_encoded, y_train)

        train_acc.append(accuracy_score(y_train, estimator.predict(X_train_encoded)))
        test_acc.append(accuracy_score(y_test, estimator.predict(X_test_encoded)))

        train_f1.append(f1_score(y_train, estimator.predict(X_train_encoded), average='macro')) # average='macro' - из за дисбаналанса классов
        test_f1.append(f1_score(y_test, estimator.predict(X_test_encoded), average='macro'))

    return train_acc, test_acc, train_f1, test_f1


def plot_dependence(
    param_name: str,
    grid: Sequence[Any] = range(2, 20),
    other_params_dict: Dict[str, Any] = {}
):
    '''
    Функция для отображения графика зависимости метрик качества
    от значения параметра c названием param_name

    Параметры:
    param_name (str): название параметра, который собираемся варьировать
    grid (Sequence[Any]): сетка значений параметра
    other_params_dict (Dict[str, Any]): словарь со значениями остальных параметров
    '''

    train_acc, test_acc, train_f1, test_f1 = get_train_and_test_mse(
        param_name, grid, other_params_dict
    )

    plot_some_lines(
        lines={
            'Тренировочная выборка': train_acc,
            'Тестовая выборка': test_acc,
        },
        xlabel=param_name,
        estrimator_range=grid,
        title=f'Зависимость Accuracy от {param_name}',
        metric_name=f'Accuracy',
    )

    plot_some_lines(
        lines={
            'Тренировочная выборка': train_f1,
            'Тестовая выборка': test_f1,
        },
        xlabel=param_name,
        estrimator_range=grid,
        title=f'Зависимость F1-меры (macro) от {param_name}',
        metric_name=f'F1-мера (macro)',
    )

plot_dependence("max_depth", [2, 5, 8, 12, 15],
                {"n_estimators": grid_search.best_params_["n_estimators"],
                 "min_samples_split": grid_search.best_params_["min_samples_split"]})
```

```python
X = data.drop(columns=["RainTomorrow"])
y = data["RainTomorrow"]

compass_order = [
        'N', 'NNE', 'NE', 'ENE',
        'E', 'ESE', 'SE', 'SSE',
        'S', 'SSW', 'SW', 'WSW',
        'W', 'WNW', 'NW', 'NNW'
    ]
cols = ['WindGustDir', 'WindDir9am', 'WindDir3pm']

# делаем предобработку признаков которую обсудили выше
preprocessor = ColumnTransformer(
    transformers=[
        ('wind_dir',
         OrdinalEncoder(
            categories=[compass_order, compass_order, compass_order],
            handle_unknown='use_encoded_value', # unknown_value = -1, для неизвестных на момент обучения подставится, например в compass order не nan, так что...
            unknown_value=-1,
            encoded_missing_value=-1  # Nan станет -1
         ),                             # что применяем
         cols),                         # к чему применяем

        ('locations',
         TargetEncoder(target_type='auto'),
         ['Location'] # к чему применяем
         )
    ]
)

best_params = {'max_depth': 15, 'min_samples_split': 3, 'n_estimators': 50}
best_model = RandomForestClassifier(# random_state=RANDOM_STATE,
                               max_depth=best_params['max_depth'],
                               min_samples_split=best_params['min_samples_split'],
                               n_estimators=best_params['n_estimators'],
                               class_weight='balanced'
)
best_pipeline = Pipeline(steps=[
    ('preprocessor', preprocessor),
    ('classifier', best_model)]
                        )

holdout_results = []
kfold_mean_results = []
kfold_var_results = []
shuffle_split_mean_results = []
shuffle_split_var_results = []

holdout_results_strat = []
kfold_mean_results_strat = []
kfold_var_results_strat = []
shuffle_split_mean_results_strat = []
shuffle_split_var_results_strat = []

for i in tqdm(range(100)):
    # 1. перемешиваем данные
    shuffled_idx = np.random.permutation(len(X))
    X_shuffled = X.iloc[shuffled_idx].reset_index(drop=True)
    y_shuffled = y.iloc[shuffled_idx].reset_index(drop=True)
    # делаем как на семинаре, без тест

    # ========= Теперь собсвтенно запускаем разные CV (Cross-Val Strategies) и считаем метрики
    # ====== Делаем 2. однократное разбиение
    X_train, X_val, y_train, y_val = train_test_split(
        X_shuffled, y_shuffled,
        test_size=0.2, shuffle=False
    )

    # Результаты 2
    best_pipeline.fit(X_train, y_train)
    holdout_results += [
        f1_score(y_val, best_pipeline.predict(X_val), average='macro')
    ]

    # 3. KFold
    kfold = KFold(
        n_splits=5,       # Количество фолдов
        shuffle=False,    # уже сами пошафлили
    )
    kfold_results = cross_val_score(
        estimator=best_pipeline,    # Модель, качество которой хотим оценить
        X=X_shuffled,          # Данные для обучения (не содержат целевую переменную)
        y=y_shuffled,          # Значения целевой переменной
        cv=kfold,           # Стратегия кросс-валидации
        # Метрика качества, она будет максимизироваться
        scoring=make_scorer(fbeta_score, beta=1, average='macro'),
        n_jobs=N_JOBS,      # Количество ядер для вычислений, -1 - использование всех ядер
    )
    kfold_mean_results += [np.mean(kfold_results)]
    kfold_var_results += [np.var(kfold_results)]

    # 4. Shuffle Split
    shuffle_split = ShuffleSplit(
        n_splits=5,       # Количество итераций перемешивания с разбиением на train и test
        test_size=0.2,    # Доля объектов, которые хотим класть в test на каждой итерации
        # random_state=RANDOM_STATE,  # Фиксируем случайность
    )
    shuffle_split_results = cross_val_score(
        estimator=best_pipeline,    # Модель, качество которой хотим оценить
        X=X_shuffled,          # Данные для обучения (не содержат целевую переменную)
        y=y_shuffled,          # Значения целевой переменной
        cv=shuffle_split,   # Стратегия кросс-валидации
        # Метрика качества, она будет максимизироваться
        scoring=make_scorer(fbeta_score, beta=1, average='macro'),
        n_jobs=N_JOBS,      # Количество ядер для вычислений, -1 - использование всех ядер
    )
    shuffle_split_mean_results += [np.mean(shuffle_split_results)]
    shuffle_split_var_results += [np.var(shuffle_split_results)]

    # Делаем 2-4 для стартифицированных СV (Cross-Val Strategies)
    # По идее, стратифицированное должно работать лучше на наших данных, так как у нас в таргете дисбаланс
    # Стратифицированное Однократное разбиение
    X_train, X_val, y_train, y_val = train_test_split(
        X_shuffled, y_shuffled,
        test_size=0.2, shuffle=True, # shuffle=True без этого стратификация падает с ошибкой
        stratify=y_shuffled # стратификация
    )
    best_pipeline.fit(X_train, y_train)

    holdout_results_strat += [
        f1_score(y_val, best_pipeline.predict(X_val), average="macro")
    ]

    # Стратифицированный KFold
    start_kfold = StratifiedKFold(n_splits=5, shuffle=False)
    start_kfold_results = cross_val_score(
        best_pipeline,
        X_shuffled, y_shuffled,
        cv=start_kfold,
        scoring=make_scorer(fbeta_score, beta=1, average='macro'),
        n_jobs=N_JOBS
    )
    kfold_mean_results_strat += [np.mean(start_kfold_results)]
    kfold_var_results_strat += [np.var(start_kfold_results)]

    # Стратифицированный ShuffleSplit
    strat_shuffle_split = StratifiedShuffleSplit(n_splits=5, test_size=0.2)
    strat_shuffle_split_results = cross_val_score(
        best_pipeline,
        X_shuffled, y_shuffled,
        cv=strat_shuffle_split,
        scoring=make_scorer(fbeta_score, beta=1, average='macro'),
        n_jobs=N_JOBS
    )

    shuffle_split_mean_results_strat += [np.mean(strat_shuffle_split_results)]
    shuffle_split_var_results_strat += [np.var(strat_shuffle_split_results)]
```

Лично мне, Kfold кажется "честнее"/"надёжнее", чем Shiffle Split, поскольку Shuffle Split может в обучающую выборку несколько раз отправлять одни и те же шумные объекты, которые "собъют модель с толку", и немножко увеличат дисперсию (чуть менее устойчивой станет модель), за счет чего могут быть проблемки с корректными предсказаниями и значение f1-меры среднее как раз будет чуть поменьше. Но опять же, я привела худший случай, а Shuffle Split рандомная вещь, может и хорошо отработать. Вообще, мне кажется, что если можно обойтись без случайности (в контексте методов кросс валидации, то лучше это сделать - и использовать "честный" Kfold, в котором каждый объект однажды побывает как в качесвте обучающего так и в качестве валидационного).

Однако у Shuffle Split есть и преимущества:

* Является хорошей альтернативой KFold, так как дает более четкий контроль над количеством итераций и разбиением на train и val. В KFold соотношение между размерами train и val жестко связано с количеством итераций.
* Результат разбиения случаен, поэтому не зависит от порядка объектов в данных.
* Позволяет выполнить большее количество итераций, чем KFold, что может быть полезно для более стабильной оценки качества.
